# TCGA-BRCA Clinical Biotab Review

This notebook reviews parsed TCGA-BRCA Clinical Supplement biotab tables from disk only.
It is validation-oriented and remains part of source audit rather than cohort construction.

## Load the latest saved clinical biotab parse run

```python
from __future__ import annotations

import json
from pathlib import Path

import pandas as pd
from IPython.display import display


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / ".git").exists():
            return candidate
    raise FileNotFoundError("Could not locate the repository root from the current working directory.")


repo_root = find_repo_root(Path.cwd())
latest_pointer_path = (
    repo_root / "01-data" / "audit" / "tcga-brca" / "variables" / "tcga_brca_clinical_biotabs_latest.json"
)
if not latest_pointer_path.exists():
    raise FileNotFoundError(
        f"Latest clinical biotab pointer not found: {latest_pointer_path}. Run the parsing script first."
    )

latest_pointer = json.loads(latest_pointer_path.read_text(encoding="utf-8"))
manifest_path = repo_root / latest_pointer["table_manifest_tsv"]
run_log_path = repo_root / latest_pointer["run_log_json"]
results_root = repo_root / "09-trials" / "01-tcga-only-source-audited" / "05-results"
results_root.mkdir(parents=True, exist_ok=True)

display(pd.DataFrame([latest_pointer]))
```

## Load the saved manifest and run log

```python
manifest_df = pd.read_csv(manifest_path, sep="\t")
run_log = json.loads(run_log_path.read_text(encoding="utf-8"))

print(f"Manifest TSV: {manifest_path}")
print(f"Run log: {run_log_path}")
print(f"Tables listed: {len(manifest_df)}")
display(manifest_df)
```

## Table inventory, source-file mapping, and shape checks

```python
parsed_tables = {}
summary_rows = []

for row in manifest_df.to_dict(orient="records"):
    table_path = repo_root / row["output_path"]
    table_df = pd.read_csv(table_path, sep="\t", dtype=str, keep_default_na=False, na_filter=False)
    parsed_tables[row["table_name"]] = table_df
    summary_rows.append(
        {
            "parse_run_id": row["parse_run_id"],
            "source_run_id": row["source_run_id"],
            "table_name": row["table_name"],
            "source_filename": row["source_filename"],
            "output_filename": row["output_filename"],
            "row_count": int(len(table_df)),
            "column_count": int(len(table_df.columns)),
            "manifest_row_count": int(row["row_count"]),
            "manifest_column_count": int(row["column_count"]),
            "row_count_matches_manifest": bool(len(table_df) == int(row["row_count"])),
            "column_count_matches_manifest": bool(len(table_df.columns) == int(row["column_count"])),
            "source_path": row["source_path"],
            "output_path": row["output_path"],
        }
    )

summary_df = pd.DataFrame(summary_rows).sort_values("table_name").reset_index(drop=True)
summary_path = results_root / "15_clinical_biotab_table_summary.tsv"
summary_df.to_csv(summary_path, sep="\t", index=False)

print(f"Saved: {summary_path}")
display(summary_df)
```

## Column-name review and field-group presence

This section summarizes major field groups by column name only and does not interpret the values.

```python
FIELD_GROUP_PATTERNS = {
    "demographics": ["gender", "race", "ethnicity", "menopause", "birth", "age"],
    "stage": ["ajcc", "stage", "clinical_t", "clinical_n", "clinical_m", "pathologic", "diagnosis"],
    "receptor": ["er_", "pr_", "her2", "receptor"],
    "drug": ["drug", "pharm", "pharmaceutical", "regimen", "therapy_type"],
    "radiation": ["radiation", "rt_", "irradiat"],
    "follow_up": ["follow_up", "followup", "last_contact", "vital_status", "death_days", "days_to_death"],
    "new_tumor_event": ["new_tumor_event", "new_neoplasm", "nte_"],
    "pathology": ["histolog", "margin", "lymph", "tumor_status", "subdivision"],
}

field_group_rows = []
for table_name, table_df in parsed_tables.items():
    column_names = list(table_df.columns)
    lower_names = [name.lower() for name in column_names]
    for field_group, patterns in FIELD_GROUP_PATTERNS.items():
        matched_columns = [
            column_names[index]
            for index, lower_name in enumerate(lower_names)
            if any(pattern in lower_name for pattern in patterns)
        ]
        field_group_rows.append(
            {
                "table_name": table_name,
                "field_group": field_group,
                "matched_column_count": len(matched_columns),
                "matched_columns": json.dumps(matched_columns, ensure_ascii=True),
            }
        )

field_group_df = pd.DataFrame(field_group_rows).sort_values(["table_name", "field_group"]).reset_index(drop=True)
field_group_path = results_root / "16_clinical_biotab_field_group_presence.tsv"
field_group_df.to_csv(field_group_path, sep="\t", index=False)

print(f"Saved: {field_group_path}")
display(field_group_df)
```

## Missingness-like token review

This section counts common placeholder tokens directly from the parsed tables without changing the source values.

```python
MISSING_LIKE_TOKENS = {
    "blank": "",
    "not_available": "[Not Available]",
    "not_applicable": "[Not Applicable]",
    "unknown": "[Unknown]",
    "not_evaluated": "[Not Evaluated]",
    "discrepancy": "[Discrepancy]",
}

missingness_rows = []
for table_name, table_df in parsed_tables.items():
    row_count = len(table_df)
    for column_name in table_df.columns:
        values = table_df[column_name].astype(str)
        token_counts = {label: int((values == token).sum()) for label, token in MISSING_LIKE_TOKENS.items()}
        missing_like_count = sum(token_counts.values())
        missingness_rows.append(
            {
                "table_name": table_name,
                "column_name": column_name,
                "row_count": int(row_count),
                "blank_count": token_counts["blank"],
                "not_available_count": token_counts["not_available"],
                "not_applicable_count": token_counts["not_applicable"],
                "unknown_count": token_counts["unknown"],
                "not_evaluated_count": token_counts["not_evaluated"],
                "discrepancy_count": token_counts["discrepancy"],
                "missing_like_count": missing_like_count,
                "missing_like_fraction": float(missing_like_count / row_count) if row_count else 0.0,
            }
        )

missingness_df = (
    pd.DataFrame(missingness_rows)
    .sort_values(["table_name", "missing_like_fraction", "column_name"], ascending=[True, False, True])
    .reset_index(drop=True)
)
missingness_path = results_root / "17_clinical_biotab_missingness_summary.tsv"
missingness_df.to_csv(missingness_path, sep="\t", index=False)

print(f"Saved: {missingness_path}")
display(missingness_df.head(25))
```
