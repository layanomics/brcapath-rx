# TCGA-BRCA Biospecimen Biotab Review

This notebook reviews parsed TCGA-BRCA core Biospecimen Supplement biotab tables from disk only.
It is validation-oriented and remains part of source audit rather than cohort construction.


## Load the latest saved biospecimen biotab parse run


```python
from __future__ import annotations

import json
from pathlib import Path

import pandas as pd
from IPython.display import display


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / ".git").exists():
            return candidate
    raise FileNotFoundError("Could not locate the repository root from the current working directory.")


repo_root = find_repo_root(Path.cwd())
latest_pointer_path = (
    repo_root / "01-data" / "audit" / "tcga-brca" / "variables" / "tcga_brca_biospecimen_biotabs_latest.json"
)
if not latest_pointer_path.exists():
    raise FileNotFoundError(
        f"Latest biospecimen biotab pointer not found: {latest_pointer_path}. Run the parsing script first."
    )

latest_pointer = json.loads(latest_pointer_path.read_text(encoding="utf-8"))
manifest_path = repo_root / latest_pointer["table_manifest_tsv"]
source_inventory_path = repo_root / latest_pointer["source_inventory_tsv"]
run_log_path = repo_root / latest_pointer["run_log_json"]
results_root = repo_root / "09-trials" / "01-tcga-only-source-audited" / "05-results"
results_root.mkdir(parents=True, exist_ok=True)

display(pd.DataFrame([latest_pointer]))
```

## Load the saved source inventory, manifest, and run log

The source inventory should show the discovered core biospecimen biotabs plus the deferred `ssf_*.txt` text tables.


```python
source_inventory_df = pd.read_csv(source_inventory_path, sep="\t", dtype=str, keep_default_na=False, na_filter=False)
manifest_df = pd.read_csv(manifest_path, sep="\t", dtype=str, keep_default_na=False, na_filter=False)
run_log = json.loads(run_log_path.read_text(encoding="utf-8"))

print(f"Source inventory TSV: {source_inventory_path}")
print(f"Manifest TSV: {manifest_path}")
print(f"Run log: {run_log_path}")
print(f"Discovered biospecimen biotab text files: {len(source_inventory_df)}")
print(f"Parsed tables listed: {len(manifest_df)}")
display(source_inventory_df)
display(manifest_df)
```

## Table inventory, source-file mapping, and shape checks


```python
parsed_tables = {}
schema_tables = {}
summary_rows = []

for row in manifest_df.to_dict(orient="records"):
    table_path = repo_root / row["output_path"]
    schema_path = repo_root / row["schema_path"]
    table_df = pd.read_csv(table_path, sep="\t", dtype=str, keep_default_na=False, na_filter=False)
    schema_df = pd.read_csv(schema_path, sep="\t", dtype=str, keep_default_na=False, na_filter=False)
    parsed_tables[row["table_name"]] = table_df
    schema_tables[row["table_name"]] = schema_df
    summary_rows.append(
        {
            "parse_run_id": row["parse_run_id"],
            "source_run_id": row["source_run_id"],
            "table_name": row["table_name"],
            "source_filename": row["source_filename"],
            "output_filename": row["output_filename"],
            "row_count": int(len(table_df)),
            "column_count": int(len(table_df.columns)),
            "manifest_row_count": int(row["row_count"]),
            "manifest_column_count": int(row["column_count"]),
            "row_count_matches_manifest": bool(len(table_df) == int(row["row_count"])),
            "column_count_matches_manifest": bool(len(table_df.columns) == int(row["column_count"])),
            "header_preamble_rows": int(row["header_preamble_rows"]),
            "data_start_line": int(row["data_start_line"]),
            "source_path": row["source_path"],
            "output_path": row["output_path"],
            "schema_path": row["schema_path"],
        }
    )

summary_df = pd.DataFrame(summary_rows).sort_values("table_name").reset_index(drop=True)
summary_path = results_root / "37_biospecimen_biotab_table_summary.tsv"
summary_df.to_csv(summary_path, sep="\t", index=False)

print(f"Saved: {summary_path}")
display(summary_df)
```

## Identifier-like field groups by column name only

This section summarizes column-name groups only. It does not harmonize identifiers or interpret biology.


```python
IDENTIFIER_GROUP_PATTERNS = {
    "patient": ["patient"],
    "sample": ["sample"],
    "portion": ["portion"],
    "analyte": ["analyte"],
    "slide": ["slide"],
    "aliquot": ["aliquot"],
    "barcode": ["barcode"],
    "uuid": ["uuid"],
    "protocol": ["protocol"],
    "shipment": ["shipment"],
}

identifier_group_rows = []
for table_name, table_df in parsed_tables.items():
    column_names = list(table_df.columns)
    lower_names = [name.lower() for name in column_names]
    for identifier_group, patterns in IDENTIFIER_GROUP_PATTERNS.items():
        matched_columns = [
            column_names[index]
            for index, lower_name in enumerate(lower_names)
            if any(pattern in lower_name for pattern in patterns)
        ]
        identifier_group_rows.append(
            {
                "table_name": table_name,
                "identifier_group": identifier_group,
                "matched_column_count": len(matched_columns),
                "matched_columns": json.dumps(matched_columns, ensure_ascii=True),
            }
        )

identifier_group_df = (
    pd.DataFrame(identifier_group_rows)
    .sort_values(["table_name", "identifier_group"])
    .reset_index(drop=True)
)
identifier_group_path = results_root / "38_biospecimen_biotab_identifier_field_groups.tsv"
identifier_group_df.to_csv(identifier_group_path, sep="\t", index=False)

print(f"Saved: {identifier_group_path}")
display(identifier_group_df)
```

## Missingness-like token review

This section counts common placeholder tokens directly from the parsed tables without changing the source values.


```python
MISSING_LIKE_TOKENS = {
    "blank": "",
    "not_available": "[Not Available]",
    "not_applicable": "[Not Applicable]",
    "unknown": "[Unknown]",
    "not_evaluated": "[Not Evaluated]",
    "discrepancy": "[Discrepancy]",
}

missingness_rows = []
for table_name, table_df in parsed_tables.items():
    row_count = len(table_df)
    for column_name in table_df.columns:
        values = table_df[column_name].astype(str)
        token_counts = {label: int((values == token).sum()) for label, token in MISSING_LIKE_TOKENS.items()}
        missing_like_count = sum(token_counts.values())
        non_missing_like_values = values.loc[~values.isin(MISSING_LIKE_TOKENS.values())]
        missingness_rows.append(
            {
                "table_name": table_name,
                "column_name": column_name,
                "row_count": int(row_count),
                "blank_count": token_counts["blank"],
                "not_available_count": token_counts["not_available"],
                "not_applicable_count": token_counts["not_applicable"],
                "unknown_count": token_counts["unknown"],
                "not_evaluated_count": token_counts["not_evaluated"],
                "discrepancy_count": token_counts["discrepancy"],
                "missing_like_count": missing_like_count,
                "missing_like_fraction": float(missing_like_count / row_count) if row_count else 0.0,
                "distinct_non_missing_like_count": int(non_missing_like_values.nunique()),
            }
        )

missingness_df = (
    pd.DataFrame(missingness_rows)
    .sort_values(["table_name", "missing_like_fraction", "column_name"], ascending=[True, False, True])
    .reset_index(drop=True)
)
missingness_path = results_root / "39_biospecimen_biotab_missingness_summary.tsv"
missingness_df.to_csv(missingness_path, sep="\t", index=False)

print(f"Saved: {missingness_path}")
display(missingness_df.head(25))
```

## Linkage-field review by field names only

This section summarizes likely linkage paths from saved headers only. It does not harmonize identifiers or define a final hierarchy.


```python
ROLE_PATTERNS = {
    "patient_fields": ["patient_barcode", "patient_uuid", "bcr_patient"],
    "sample_fields": ["sample_barcode", "sample_uuid", "bcr_sample"],
    "portion_fields": ["portion_barcode", "portion_uuid", "portion_number", "portion_sequence"],
    "analyte_fields": ["analyte_barcode", "analyte_uuid", "analyte_type", "subportion_sequence"],
    "slide_fields": ["slide_barcode", "slide_uuid", "ffpe_slide_uuid"],
    "aliquot_fields": ["aliquot_barcode", "aliquot_uuid", "biospecimen_barcode_bottom"],
    "protocol_fields": ["protocol"],
    "shipment_fields": ["shipment"],
}


def match_columns(column_names: list[str], patterns: list[str]) -> list[str]:
    matched = []
    for column_name in column_names:
        lower_name = column_name.lower()
        if any(pattern in lower_name for pattern in patterns):
            matched.append(column_name)
    return matched


linkage_rows = []
for row in summary_df.to_dict(orient="records"):
    table_name = row["table_name"]
    column_names = list(parsed_tables[table_name].columns)
    matched = {role: match_columns(column_names, patterns) for role, patterns in ROLE_PATTERNS.items()}

    likely_edges = []
    if matched["patient_fields"] and matched["sample_fields"]:
        likely_edges.append("patient -> sample")
    if matched["sample_fields"] and matched["portion_fields"]:
        likely_edges.append("sample -> portion")
    if matched["sample_fields"] and matched["analyte_fields"]:
        likely_edges.append("sample -> analyte")
    if matched["sample_fields"] and matched["slide_fields"]:
        likely_edges.append("sample -> slide")
    if matched["sample_fields"] and matched["aliquot_fields"]:
        likely_edges.append("sample -> aliquot")
    if matched["analyte_fields"] and matched["protocol_fields"]:
        likely_edges.append("analyte -> protocol")
    if matched["portion_fields"] and matched["shipment_fields"]:
        likely_edges.append("portion -> shipment")
    if matched["aliquot_fields"] and matched["shipment_fields"]:
        likely_edges.append("aliquot -> shipment")

    linkage_rows.append(
        {
            "table_name": table_name,
            "source_filename": row["source_filename"],
            "patient_fields": json.dumps(matched["patient_fields"], ensure_ascii=True),
            "sample_fields": json.dumps(matched["sample_fields"], ensure_ascii=True),
            "portion_fields": json.dumps(matched["portion_fields"], ensure_ascii=True),
            "analyte_fields": json.dumps(matched["analyte_fields"], ensure_ascii=True),
            "slide_fields": json.dumps(matched["slide_fields"], ensure_ascii=True),
            "aliquot_fields": json.dumps(matched["aliquot_fields"], ensure_ascii=True),
            "protocol_fields": json.dumps(matched["protocol_fields"], ensure_ascii=True),
            "shipment_fields": json.dumps(matched["shipment_fields"], ensure_ascii=True),
            "likely_linkage_edges": json.dumps(likely_edges, ensure_ascii=True),
            "linkage_comment": "Field-name-only linkage hints. No identifier harmonization or final hierarchy is performed here.",
        }
    )

linkage_df = pd.DataFrame(linkage_rows).sort_values("table_name").reset_index(drop=True)
linkage_path = results_root / "40_biospecimen_biotab_linkage_fields.tsv"
linkage_df.to_csv(linkage_path, sep="\t", index=False)

print(f"Saved: {linkage_path}")
print("Likely linkage path by field names only: patient -> sample -> portion / analyte / slide / aliquot, plus analyte -> protocol and shipment-linked portion or aliquot fields where present.")
display(linkage_df)
```

## Preamble and schema summary

This section confirms that the saved schema files carry source line-2 CDE metadata while the primary parsed tables preserve source line-1 headers exactly.


```python
preamble_rows = []
for row in manifest_df.to_dict(orient="records"):
    table_name = row["table_name"]
    schema_df = schema_tables[table_name]
    alternate_values = schema_df["alternate_column_name"].astype(str)
    cde_values = schema_df["cde_id_raw"].astype(str)
    cde_columns = schema_df.loc[cde_values != "", "raw_column_name"].tolist()
    preamble_rows.append(
        {
            "table_name": table_name,
            "source_filename": row["source_filename"],
            "header_preamble_rows": int(row["header_preamble_rows"]),
            "data_start_line": int(row["data_start_line"]),
            "column_count": int(row["column_count"]),
            "schema_row_count": int(len(schema_df)),
            "non_empty_alternate_column_name_count": int((alternate_values != "").sum()),
            "non_empty_cde_id_count": int((cde_values != "").sum()),
            "cde_id_columns": json.dumps(cde_columns, ensure_ascii=True),
            "parse_notes": row["parse_notes"],
        }
    )

preamble_df = pd.DataFrame(preamble_rows).sort_values("table_name").reset_index(drop=True)
preamble_path = results_root / "41_biospecimen_biotab_preamble_summary.tsv"
preamble_df.to_csv(preamble_path, sep="\t", index=False)

print(f"Saved: {preamble_path}")
display(preamble_df)
```
